```python
from dissipationtheory.constants import ureg
from dissipationtheory.constants import kb, Troom, qe, epsilon0
import numpy as np
```

# Reproduce Roger Loring's Mathematica calculation of friction

Cantilever and sample properties

```python
R_c = ureg.Quantity(35, 'nm') # cantilever radius
z_c = ureg.Quantity(75, 'nm') # tip-charge separation
V_ts = ureg.Quantity(1, 'V')  # tip voltage

omega_c = 2 * np.pi * ureg.Quantity(75, 'kHz') # cantilever frequency 
e_r = ureg.Quantity(complex(20, -0.02), '')    # sample dielectric constant  
```

Derived quantities

```python
q_0 = 4 * np.pi * epsilon0 * R_c * V_ts  # tip charge
xi = (e_r - 1)/(e_r + 1)                # dielectric-constant factor
```

Compute the friction

```python
gamma = - (q_0**2 * xi.imag)/(32 * np.pi * epsilon0 * omega_c * z_c**3)
gamma.to('pN s/m')
```

# Reproduce the low-$\rho$ limit of the friction in Fig 06 of Study 77

Cantilever and sample properties

```python
R_c = ureg.Quantity(35, 'nm') # cantilever radius
h = ureg.Quantity(75, 'nm')   # tip-sample separation
V_ts = ureg.Quantity(1, 'V')  # tip voltage

omega_c = 2 * np.pi * ureg.Quantity(75, 'kHz') # cantilever frequency 
e_r = ureg.Quantity(complex(2.8, -0.2), '')    # sample dielectric constant  
```

Derived quantities

```python
q_0 = 4 * np.pi * epsilon0 * R_c * V_ts  # tip charge
xi = (e_r - 1)/(e_r + 1)                # dielectric-constant factor
z_c = h + R_c                           # tip-sample separation
```

Compute the friction

```python
gamma = - (q_0**2 * xi.imag)/(32 * np.pi * epsilon0 * omega_c * z_c**3)
gamma.to('pN s/m')
```
